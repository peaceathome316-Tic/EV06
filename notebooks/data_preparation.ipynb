# Preparación, limpieza y estructuración de datos con Pandas

## Proyecto: Accidentalidad vial en Bogotá

**Actividad:** GA1-220501094-01-AA1-EV07

### Pregunta de análisis

¿A qué horas, en qué días de la semana y en qué localidades de Bogotá ocurrieron más siniestros viales con heridos o muertos durante 2025, y qué clases de vehículo estuvieron más involucradas?

### Objetivo

Preparar, limpiar y estructurar los datos de accidentalidad vial de Bogotá correspondientes a 2025 mediante Python y Pandas, garantizando su calidad y consistencia para realizar análisis posteriores.

### Fuentes de información

- `anuario_siniestros.csv`: información general de los siniestros.
- `anuario_vehiculos.csv`: vehículos involucrados.
- `anuario_actor_vial.csv`: información de las personas involucradas y su condición.
- `accidentes.csv`: fuente complementaria con información histórica de víctimas.

### Metodología

1. Cargar y explorar las fuentes de datos.
2. Identificar y tratar valores faltantes, duplicados e inconsistencias.
3. Estandarizar formatos y tipos de datos.
4. Filtrar y analizar los siniestros de 2025 con heridos o muertos.
5. Agrupar la información por hora, día, localidad y clase de vehículo.
6. Relacionar las tablas mediante identificadores comunes.
7. Validar y exportar los conjuntos de datos procesados.

```python
# Importar librerías necesarias

import pandas as pd
import numpy as np
from pathlib import Path

# Configuración para visualizar todas las columnas
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.max_colwidth', None)

# Definir rutas del proyecto
RUTA_DATOS = Path("../data/raw")
RUTA_PROCESADOS = Path("../data/processed")

# Crear la carpeta de procesados si no existe
RUTA_PROCESADOS.mkdir(parents=True, exist_ok=True)

print("Librerías importadas correctamente.")
print("Ruta de datos originales:", RUTA_DATOS.resolve())
print("Ruta de datos procesados:", RUTA_PROCESADOS.resolve())
```

```python
# Función para identificar valores faltantes

def revisar_valores_faltantes(df, nombre):
    resultado = pd.DataFrame({
        "Valores faltantes": df.isnull().sum(),
        "Porcentaje (%)": (df.isnull().mean() * 100).round(2)
    })

    resultado = resultado[resultado["Valores faltantes"] > 0]
    resultado = resultado.sort_values(
        by="Porcentaje (%)",
        ascending=False
    )

    print(f"\nDIAGNÓSTICO DE VALORES FALTANTES: {nombre}")
    if resultado.empty:
        print("No se encontraron valores faltantes.")
    else:
        display(resultado)

# Revisar las cuatro fuentes

revisar_valores_faltantes(df_siniestros, "Siniestros")
revisar_valores_faltantes(df_vehiculos, "Vehículos")
revisar_valores_faltantes(df_actores, "Actores viales")
revisar_valores_faltantes(df_accidentes, "Accidentes históricos")
```

```python
# Identificar registros duplicados en cada fuente

def revisar_duplicados(df, nombre):
    duplicados = df.duplicated().sum()

    print(f"{nombre}:")
    print(f"Total de registros: {len(df)}")
    print(f"Registros duplicados: {duplicados}")
    print("-" * 40)

revisar_duplicados(df_siniestros, "Siniestros")
revisar_duplicados(df_vehiculos, "Vehículos")
revisar_duplicados(df_actores, "Actores viales")
revisar_duplicados(df_accidentes, "Accidentes históricos")
```

```python
# Revisar los tipos de datos de las cuatro fuentes

def revisar_tipos_datos(df, nombre):
    print(f"\nTIPOS DE DATOS: {nombre}")
    display(pd.DataFrame({
        "Tipo de dato": df.dtypes.astype(str),
        "Valores únicos": df.nunique(),
        "Valores faltantes": df.isnull().sum()
    }))

revisar_tipos_datos(df_siniestros, "Siniestros")
revisar_tipos_datos(df_vehiculos, "Vehículos")
revisar_tipos_datos(df_actores, "Actores viales")
revisar_tipos_datos(df_accidentes, "Accidentes históricos")
```

```python
# Validación general de las tablas limpias

tablas = {
    "Siniestros": (df_siniestros, df_siniestros_limpio),
    "Vehículos": (df_vehiculos, df_vehiculos_limpio),
    "Actores viales": (df_actores, df_actores_limpio)
}

for nombre, (original, limpio) in tablas.items():
    print(f"\n--- {nombre} ---")

    print("Dimensiones originales:", original.shape)
    print("Dimensiones limpias:", limpio.shape)

    print("Filas duplicadas originales:", original.duplicated().sum())
    print("Filas duplicadas limpias:", limpio.duplicated().sum())

    assert original.shape == limpio.shape, (
        f"Error: cambiaron las dimensiones de {nombre}"
    )

print("\nValidación completada correctamente.")
```

```python
# Comprobar la relación entre las tres tablas

ids_siniestros = set(df_siniestros_limpio["Codigo_Siniestro"].dropna())
ids_vehiculos = set(df_vehiculos_limpio["Codigo_Siniestro"].dropna())
ids_actores = set(df_actores_limpio["Codigo_Siniestro"].dropna())

print("Siniestros únicos:", len(ids_siniestros))
print("Códigos en vehículos:", len(ids_vehiculos))
print("Códigos en actores:", len(ids_actores))

print("\nVehículos sin siniestro asociado:", len(ids_vehiculos - ids_siniestros))
print("Actores sin siniestro asociado:", len(ids_actores - ids_siniestros))
print("Siniestros sin vehículos:", len(ids_siniestros - ids_vehiculos))
print("Siniestros sin actores:", len(ids_siniestros - ids_actores))
```

```python
# Filtrar siniestros con heridos o muertos

df_siniestros_interes = (
    df_siniestros_limpio
    .query("Gravedad_indicador_30d in ['Con Heridos', 'Con Muertos']")
    .copy()
)

print("Total de siniestros originales:", len(df_siniestros_limpio))
print("Siniestros con heridos o muertos:", len(df_siniestros_interes))

print("\nDistribución por gravedad:")
display(
    df_siniestros_interes["Gravedad_indicador_30d"]
    .value_counts()
)

# Validar el filtro
assert df_siniestros_interes["Gravedad_indicador_30d"].isin(
    ["Con Heridos", "Con Muertos"]
).all()

print("\nFiltro validado correctamente.")
```

```python
# Cantidad de siniestros por hora

siniestros_por_hora = (
    df_siniestros_interes
    .groupby("Hora_Siniestro")
    .agg(
        Total_Siniestros=("Codigo_Siniestro", "count")
    )
    .sort_values("Total_Siniestros", ascending=False)
    .reset_index()
)

display(siniestros_por_hora)

print("\nLas 5 horas con más siniestros:")
display(siniestros_por_hora.head(5))
```

```python
# Cantidad de siniestros por día de la semana

siniestros_por_dia = (
    df_siniestros_interes
    .groupby("Dia_Semana_Siniestro")
    .agg(
        Total_Siniestros=("Codigo_Siniestro", "count")
    )
    .sort_values("Total_Siniestros", ascending=False)
    .reset_index()
)

display(siniestros_por_dia)
```

```python
# Cantidad de siniestros por localidad

siniestros_por_localidad = (
    df_siniestros_interes
    .groupby("Localidad")
    .agg(
        Total_Siniestros=("Codigo_Siniestro", "count")
    )
    .sort_values("Total_Siniestros", ascending=False)
    .reset_index()
)

display(siniestros_por_localidad)

print("\nLas 5 localidades con más siniestros:")
display(siniestros_por_localidad.head(5))
```

```python
# Filtrar vehículos asociados a los siniestros de interés

ids_siniestros_interes = df_siniestros_interes["Codigo_Siniestro"].unique()

df_vehiculos_interes = df_vehiculos_limpio[
    df_vehiculos_limpio["Codigo_Siniestro"].isin(ids_siniestros_interes)
].copy()

# Agrupar por clase de vehículo

vehiculos_por_clase = (
    df_vehiculos_interes
    .groupby("Clase", dropna=False)
    .agg(
        Total_Vehiculos=("Codigo_Siniestro", "count")
    )
    .sort_values("Total_Vehiculos", ascending=False)
    .reset_index()
)

display(vehiculos_por_clase)

print("\nLas 5 clases de vehículos más involucradas:")
display(vehiculos_por_clase.head(5))
```

```python
# Resumir vehículos por código de siniestro

resumen_vehiculos = (
    df_vehiculos_interes
    .groupby("Codigo_Siniestro")
    .agg(
        Total_Vehiculos=("Codigo_Siniestro", "size"),
        Clases_Vehiculos=(
            "Clase",
            lambda x: ", ".join(sorted(x.dropna().unique()))
        )
    )
    .reset_index()
)

print("Dimensiones del resumen:", resumen_vehiculos.shape)
display(resumen_vehiculos.head())
```

```python
# Filtrar actores asociados a los siniestros de interés

df_actores_interes = df_actores_limpio[
    df_actores_limpio["Codigo_Siniestro"].isin(ids_siniestros_interes)
].copy()

# Resumir actores por código de siniestro

resumen_actores = (
    df_actores_interes
    .groupby("Codigo_Siniestro")
    .agg(
        Total_Actores=("Codigo_Siniestro", "size"),
        Total_Heridos=(
            "Gravedad_Indicador_30d",
            lambda x: (x == "HERIDO").sum()
        ),
        Total_Muertos=(
            "Gravedad_Indicador_30d",
            lambda x: (x == "MUERTO").sum()
        ),
        Condiciones_Actores=(
            "Condicion",
            lambda x: ", ".join(sorted(x.dropna().unique()))
        )
    )
    .reset_index()
)

print("Dimensiones del resumen:", resumen_actores.shape)
display(resumen_actores.head())
```

```python
# Integrar las tres tablas

df_integrado = (
    df_siniestros_interes
    .merge(
        resumen_vehiculos,
        on="Codigo_Siniestro",
        how="left",
        validate="one_to_one"
    )
    .merge(
        resumen_actores,
        on="Codigo_Siniestro",
        how="left",
        validate="one_to_one"
    )
)

print("Dimensiones del conjunto integrado:", df_integrado.shape)

print("\nPrimeros registros:")
display(df_integrado.head())

print("\nValores nulos en las columnas integradas:")
display(
    df_integrado[
        [
            "Total_Vehiculos",
            "Clases_Vehiculos",
            "Total_Actores",
            "Total_Heridos",
            "Total_Muertos",
            "Condiciones_Actores"
        ]
    ].isna().sum()
)

# Validaciones
assert len(df_integrado) == len(df_siniestros_interes)
assert df_integrado["Codigo_Siniestro"].is_unique

print("\nIntegración validada correctamente.")
```

```python
# Resumen general del conjunto integrado

print("Total de siniestros:", len(df_integrado))

print("\nTotal de vehículos involucrados:")
print(df_integrado["Total_Vehiculos"].sum())

print("\nTotal de actores involucrados:")
print(df_integrado["Total_Actores"].sum())

print("\nTotal de heridos registrados en actores:")
print(df_integrado["Total_Heridos"].sum())

print("\nTotal de muertos registrados en actores:")
print(df_integrado["Total_Muertos"].sum())

print("\nDistribución de gravedad de los siniestros:")
display(
    df_integrado["Gravedad_indicador_30d"].value_counts()
)
```

```python
# Funciones reutilizables para la preparación de datos

def limpiar_nombres_columnas(df):
    """Elimina espacios innecesarios de los nombres de columnas."""
    df = df.copy()
    df.columns = df.columns.str.strip()
    return df

def limpiar_texto(df, columnas):
    """Elimina espacios y estandariza campos de texto."""
    df = df.copy()

    for columna in columnas:
        df[columna] = (
            df[columna]
            .astype("string")
            .str.strip()
        )

    return df

def estandarizar_indicadores(df, columnas):
    """Estandariza indicadores y reemplaza nulos por NO."""
    df = df.copy()

    for columna in columnas:
        df[columna] = (
            df[columna]
            .astype("string")
            .str.strip()
            .str.upper()
            .fillna("NO")
        )

    return df

def convertir_fecha(df, columna):
    """Convierte una columna a formato fecha."""
    df = df.copy()
    df[columna] = pd.to_datetime(
        df[columna],
        errors="coerce"
    )
    return df

def convertir_coordenadas(df, columnas):
    """Convierte coordenadas a valores numéricos."""
    df = df.copy()

    for columna in columnas:
        df[columna] = pd.to_numeric(
            df[columna]
            .astype("string")
            .str.replace(",", ".", regex=False),
            errors="coerce"
        )

    return df

print("Funciones de limpieza definidas correctamente.")
```

```python
# Columnas de texto de la tabla de siniestros
columnas_texto_siniestros = [
    "Formulario",
    "MM_Siniestro",
    "Dia_Semana_Siniestro",
    "Localidad",
    "Clase_Siniestro",
    "Elemento_Choque",
    "Tipo_Objeto_Fijo",
    "Gravedad_indicador_30d"
]

# Aplicar las funciones de limpieza utilizando pipe()
df_siniestros_pipe = (
    df_siniestros
    .pipe(limpiar_nombres_columnas)
    .pipe(limpiar_texto, columnas=columnas_texto_siniestros)
    .pipe(convertir_fecha, columna="Fecha_Siniestro")
    .pipe(
        convertir_coordenadas,
        columnas=["Longitud", "Latitud"]
    )
    .pipe(
        estandarizar_indicadores,
        columnas=columnas_indicadores
    )
)

# Comprobar el resultado
print("Dimensiones:", df_siniestros_pipe.shape)

print("\nTipos de datos:")
display(df_siniestros_pipe.dtypes)

print("\nValores nulos en indicadores:")
display(
    df_siniestros_pipe[columnas_indicadores].isna().sum()
)

assert df_siniestros_pipe.shape == df_siniestros.shape
assert df_siniestros_pipe[columnas_indicadores].isna().sum().sum() == 0

print("\nPipeline de limpieza ejecutado correctamente.")
```
